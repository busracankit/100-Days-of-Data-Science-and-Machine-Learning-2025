```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('7-cyber_attack_data.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
X = df.drop("attack_type", axis=1)
y = df["attack_type"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
score = accuracy_score(y_pred, y_test)
print("score:", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: /n", confusion_matrix(y_pred, y_test))
```

```python
# one vs rest
```

```python
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier
```

```python
onevsonemodel = OneVsOneClassifier(LogisticRegression())
```

```python
onevsrestmodel =OneVsRestClassifier(LogisticRegression())
```

```python
onevsonemodel.fit(X_train, y_train)
y_pred = onevsonemodel.predict(X_test)
score = accuracy_score(y_pred, y_test)
print("score:", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: /n", confusion_matrix(y_pred, y_test))
```

```python
onevsrestmodel.fit(X_train, y_train)
y_pred = onevsonemodel.predict(X_test)
score = accuracy_score(y_pred, y_test)
print("score:", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: /n", confusion_matrix(y_pred, y_test))
```
